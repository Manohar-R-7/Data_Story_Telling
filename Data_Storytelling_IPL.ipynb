```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv("/content/matches.csv")
df.head()
```

Dataset Introduction

The IPL dataset contains match information from the Indian Premier League between 2008 and 2022. It includes details such as teams, venues, toss winners, match winners, and player awards. This dataset is useful for analyzing trends, team performance, and tournament growth.

```python
print("Total Matches:", len(df))
print("Total Seasons:", df['season'].nunique())
print("Total Teams:", pd.concat([df['team1'], df['team2']]).nunique())
```

```python
matches = df.groupby('season').size().reset_index(name='Matches')

fig = px.bar(
    matches,
    x='season',
    y='Matches',
    title='Matches Per Season'
)

fig.show()
```

```python
wins = df['winner'].value_counts().reset_index()

wins.columns = ['Team', 'Wins']

fig = px.bar(
    wins,
    x='Team',
    y='Wins',
    title='Team Wins Analysis'
)

fig.show()
```

```python
toss = df['toss_winner'].value_counts().reset_index()

toss.columns = ['Team', 'Count']

fig = px.pie(
    toss,
    names='Team',
    values='Count',
    title='Toss Winners Distribution'
)

fig.show()
```

```python
venue = df['venue'].value_counts().head(10).reset_index()

venue.columns = ['Venue', 'Matches']

fig = px.bar(
    venue,
    x='Venue',
    y='Matches',
    title='Top 10 IPL Venues'
)

fig.show()
```

Insights and Findings

1. The IPL has grown significantly over the years, with more matches being played in recent seasons.

2. Some teams consistently achieve higher win counts, indicating stronger long-term performance.

3. Winning the toss does not always result in winning the match.

4. A small number of venues host a large proportion of IPL matches.

5. The data highlights the competitive nature of the IPL across multiple seasons.

Conclusion and Recommendations

The IPL dataset reveals valuable trends related to team performance, venue usage, and tournament growth. Teams can use historical performance data to improve strategies and decision-making. Data-driven analysis helps identify strengths, weaknesses, and opportunities for improvement.
